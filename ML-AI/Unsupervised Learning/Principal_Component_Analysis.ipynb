### Task 4: Standardize the Data
---

```python
x = iris.iloc[:,0:4].values
y = iris.species.values
```

```python
from sklearn.preprocessing import StandardScaler

X = StandardScaler().fit_transform(x)
```

### Task 5: Compute the Eigenvectors and Eigenvalues
---

Covariance: $\sigma_{jk} = \frac{1}{n-1}\sum_{i=1}^{N}(x_{ij}-\bar{x_j})(x_{ik}-\bar{x_k})$

Coviance matrix: $Σ = \frac{1}{n-1}((X-\bar{x})^T(X-\bar{x}))$

```python
covariance_matrix = np.cov(X.T)

print("Covariance matrix: \n", covariance_matrix)
```

We can prove this by looking at the covariance matrix. It has the property that it is symmetric. We also constrain the each of the columns (eigenvectors) such that the values sum to one. Thus, they are orthonormal to each other.

Eigendecomposition of the covriance matrix:  $Σ = W\wedge W^{-1}$

```python
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)
print("Eigen vectors: \n", eigen_vectors, "\n")
print("Eigen vals: \n", eigen_values)
```

### Task 6: Singular Value Decomposition (SVD)
---

```python
eigen_vec_svd, s, v = np.linalg.svd(X.T)
eigen_vec_svd
```

### Task 7: Picking Principal Components Using the Explained Variance
---

```python
for val in eigen_values:
    print(val)
```

```python
variance_explained = [(i/sum(eigen_values))*100 for i in eigen_values]
variance_explained
```

```python
cumulative_variance_explained = np.cumsum(variance_explained)
cumulative_variance_explained
```

```python
sns.lineplot(x=[1,2,3,4], y=cumulative_variance_explained)
plt.xlabel("Num of components")
plt.ylabel("Cumulative explained variance")
plt.title("Explain variance vs num of components")
plt.show()
```

### Task 8: Project Data Onto Lower-Dimensional Linear Subspace
---

```python
eigen_vectors
```

```python
projection_matrix = (eigen_vectors.T[:][:])[:2].T #choosing only 1st 2 principal components 
print("Matrix : \n", projection_matrix)
```

```python
X_pca = X.dot(projection_matrix) #Xpca = X.W
```

```python
for species in ('Iris-setosa', 'Iris-versicolor', 'Iris-virginica'):
    sns.scatterplot(
        x=X_pca[y == species, 0],
        y=X_pca[y == species, 1],
        label=species
    )
```

We started off with the dataset, identified the feature variables and target classes, and then preprocessed the data. Initially, the classes were difficult to separate clearly in the original feature space.

To address this, we first **standardized the features** so that they were on a comparable scale. We then applied **Principal Component Analysis (PCA)** to reduce the dimensionality of the dataset.

In PCA, we derived the principal components by:

* Calculating the **covariance matrix** of the standardized data.
* Performing **eigendecomposition** to obtain the **eigenvalues and eigenvectors**.
* Alternatively, PCA can be computed using **Singular Value Decomposition (SVD)**.

The **eigenvalues** indicate how much variance each principal component explains. We used the **cumulative explained variance** to determine how many components to retain. By setting a threshold of **95% explained variance**, we found that the **first two principal components** captured approximately 95% of the information in the dataset.

Therefore, we selected the first two eigenvectors to form the **projection matrix**. We then projected the original standardized data onto this lower-dimensional space by taking the dot product of the data matrix with the projection matrix.

As a result, the original dataset was reduced from **4 dimensions (features) to 2 dimensions**, while retaining approximately **95% of the dataset's variance**. This lower-dimensional representation makes it easier to visualize and analyze the separation between the different classes.
